```python
import random
import numpy as np
import matplotlib.pyplot as plt
from functools import reduce
from itertools import combinations
```

O problema. Colocar 8 rainhas num tabuleiro 8×8 sem nenhuma atacar outra. Rainha ataca na mesma linha, na mesma coluna e nas diagonais.

```python
#numero rainhas (numero coordenadas)
kk = 8
```

```python
def fo(posicoes):
  soma =1
  l_p = []

  l_p = [(x,y) for x, y in combinations(posicoes, 2)]

  for posicao in l_p:
    p_um = posicao[0]
    p_dois = posicao[1]

    p_um_linha = p_um[0]
    p_um_coluna = p_um[1]
    p_dois_linha = p_dois[0]
    p_dois_coluna = p_dois[1]

    #diagonal
    if p_um_linha==p_dois_linha and p_um_coluna==p_dois_coluna:
      # print(f"{p_um, p_dois} mesma coord")
      pass
    elif abs(p_dois_linha-p_um_linha)==abs(p_dois_coluna-p_um_coluna):
      # print(f"{p_um, p_dois}Está na diagonal")
      pass
    elif p_um_linha==p_dois_linha or p_um_coluna==p_dois_coluna:
      # print(f"{p_um, p_dois}Está na mesma linha ou coluna")
      pass
    else:
      # print(f"{p_um, p_dois}Liberado")
      soma+=1



  # print(f"Maxima pontuação possível = {len(l_p)}, Alcançada = {soma}")
  
  return soma
```

```python
def ga(pop_tamanho, num_gen):

    #inicialização
    populacao = []
    fit_historico = []

    # preencher pop
    for pt in range(pop_tamanho):
        coord = random.choices([(i,j) for i in range(8) for j in range(8)],k=kk)
        populacao.append(coord)

    best_solution = None
    best_fitness = float('-inf')


    for gen in range(num_gen):

        # evaluation  (in gen)
            
        fitness = [fo(p) for p in populacao]
        feasible_pop = populacao
        
        # selection
        if feasible_pop:
            pais = random.choices(feasible_pop,weights=fitness,k=pop_tamanho)

        # crossover

        offspring = []
        numero = int(pop_tamanho/2)
        for i in range(numero-1):
            pai1,pai2,pai3,pai4 = random.choices(pais,k=4)
            # print(pai1)
            n = random.randint(1,7)
            filho = pai1[:n] + pai2[n:]
            m = random.randint(1,7)

            filho2  = pai4[:m] + pai3[m:]

        
            # print(filho)
            offspring.append(filho)
            offspring.append(filho2)
        
        # mutation
        tx = 1/8
        for i in range(len(offspring)):
            for j in range(len(offspring[i])): # [(x,y),...]
                if random.random()<tx:
                    troca = random.choices([(x,y) for x in range(8) for y in range(8)],k=1)
                    # print("Troca, ",troca[0])
                    offspring[i][j] = troca[0]
                    # offspring[i][j] = offspring[i][j]
        
        # elitism
        if best_solution is not None:
            offspring[0] = best_solution

        populacao = offspring

        feasible_solutions = populacao

        if feasible_solutions:
            best_solution = max(feasible_solutions,key= lambda lista:fo(lista))
            best_fitness = fo(best_solution)
        fit_historico.append(best_fitness)

        print(f"Geração {gen+1}, Best solution: {best_solution}, Best Fitness: {best_fitness}")
    return 'fit_historico'
```
