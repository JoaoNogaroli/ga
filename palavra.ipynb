```python
import random
import numpy as np
import matplotlib.pyplot as plt
from functools import reduce
```

#### Problema: A partir de uma lista possível de letras, fazer o algoritmo adivinhar a palavra objetivo

```python
objetivo = 'InconstiTucionaLissimamente'
# objetivo = 'EscolA'

lista_letras = 'abcdefghijklmnopqrstuvxwyzABCDEFGHIJKLMNOPQRSTUVXWYZ'

```

```python
def func_obj(palavra):
    soma=1
    for i in range(len(palavra)):
        if palavra[i] == objetivo[i]: # True false
            soma+=1
    return soma
```

```python
def ga(pop_tamanho,num_gen):

    #--------------
    #inicialização
    population = []
    fit_history = []
        #preencher a população
    for i in range(pop_tamanho):
    
        p_palavra = random.choices(lista_letras,k=len(objetivo))
        palavra = reduce(lambda x, y: x + y, p_palavra)
        
        population.append(palavra)

    best_solution = None
    best_fitness = float('-inf')

    for gen in range(num_gen):

        #---------------
        #Evaluation

        fitness = [func_obj(p) for p in population]
        feasible_population = population


        # ['lalalaal','arrrozz',]
        #---------------
        #selection
        if feasible_population:
            feasible_fitness = [func_obj(p) for p in feasible_population]
            pais = random.choices(feasible_population,weights=feasible_fitness,k=pop_tamanho)
            # pais[0] > pais[1]
            # print(pais)
        # nao precisa de else pois não há  restrições

        # ---------------
        #Crossover

        offspring = []
        for i in range(pop_tamanho):
            pai1,pai2 = random.choices(pais,k=2)
            numero_sorter = random.randint(2,len(population))
            filho = pai1[:numero_sorter]+pai2[numero_sorter:]
            offspring.append(filho)

            # ['abcdef','obriga']
            # posicao_sorteada = 2

        #------------
        # Mutation
        # taxa_mutacao = 0.1
        taxa_mutacao = 1/len(offspring)

        #filho = ['abciga']
        for i in range(len(offspring)):
            for j in range(len(offspring[i])):
                if random.random() < taxa_mutacao: #random.random() 0,1 reais
                    troca = random.choice(lista_letras)
                    posicao_troca = j
                    palavra_a_trocar = offspring[i]
                    offspring[i] = palavra_a_trocar[:posicao_troca]+troca+palavra_a_trocar[posicao_troca+1:]

         # filho = ['abciga']        -> 'abciga' -> 'abHiga'
        

        #--------------
        # Elitism   :  copiar os melhores indivíduos de uma geração diretamente para a geração seguinte,

        if best_solution is not None:
            offspring[0] = best_solution

        population = offspring

        feasible_solutions = population

        if feasible_solutions:
            best_solution = max(feasible_solutions, key= lambda palavra: func_obj(palavra))
            best_fitness = func_obj(best_solution)
        fit_history.append(best_fitness)
        print(f"Geração {gen+1}, Best solution: {best_solution}, Best Fitness: {best_fitness}")
    return 'feito'
```

```python
ga(1000,1500)
```
